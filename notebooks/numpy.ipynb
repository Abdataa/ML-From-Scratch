```python
# Random selection in numpy
mean=[0,0]
cov=[[1,2],
    [2,5]]
x=np.random.multivariate_normal(mean,cov,100)

import matplotlib.pyplot as plt
import seaborn 
seaborn.set(); # styling plot
plt.scatter(x[:,0],x[:,1]);
```

```python
indices=np.random.choice(x.shape[0],20,replace=False)
indices
```

```python
selection=x[indices] # fancy indexing to selected
```

```python
plt.scatter(x[:,0],x[:,1],color='b',alpha=0.2)
plt.scatter(selection[:,0],selection[:,1],facecolor='none',s=500)
```
